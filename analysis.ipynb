```python
plt.plot(df["Marks"])
plt.title("Student Marks Line Graph")
plt.xlabel("Student Index")
plt.ylabel("Marks")
plt.show()
```

```python
plt.bar(df["Name"][:10], df["Marks"][:10])
plt.title("Top 10 Students Marks")
plt.xlabel("Names")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.show()
```

```python
top=df.sort_values(by="Marks",ascending=False).head(25)
top.plot(
    kind="pie",
    y="Marks",
    labels=df["Name"],
    autopct="%1.1f%%",
    legend=False
)
plt.title("top 25 marks Distribution")
plt.ylabel(" ")
plt.show()
```
